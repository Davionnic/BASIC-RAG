This notebook walks through a **Retrieval-Augmented Generation** pipeline step by step.

**What RAG does:** instead of asking an LLM from memory alone, you:
1. Load your documents
2. Split them into chunks and embed them as vectors
3. When you ask a question, retrieve the most similar chunks
4. Send those chunks + your question to the LLM for a grounded answer
